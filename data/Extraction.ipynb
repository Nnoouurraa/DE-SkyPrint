```python
url = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude": 24.7136,
    "longitude": 46.6753,
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "wind_speed_10m",
        "wind_direction_10m",
        "surface_pressure"
    ]
}

response = requests.get(url, params=params, timeout=30)

print("Status:", response.status_code)

data = response.json()

print(data.keys())
```
